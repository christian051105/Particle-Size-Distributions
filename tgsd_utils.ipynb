```python
# tgsd_utils.py
import numpy as np

def shannon_entropy(p, base=2):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]            # 0*log(0) := 0 by convention
    p = p / p.sum()          # ensure normalization
    return -np.sum(p * np.log(p) / np.log(base))

def normalized_entropy(p, base=2):
    """H / H_max, so comparable across different bin counts."""
    p = np.asarray(p, dtype=float)
    n_nonzero = np.sum(p > 0)
    H = shannon_entropy(p, base=base)
    H_max = np.log(n_nonzero) / np.log(base)
    return H / H_max if H_max > 0 else np.nan
```
